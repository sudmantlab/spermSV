## Examining filtered VCF ALT sequences -> `repeatmasker` `.gff` output 

````python
repeat_gff = pd.read_table("../../output_t2t/sniffles/mosaic/multi_sample/repeatmasker/multi_sample.filtered.fa.out.gff", skiprows = 3, header = None)
repeat_gff.columns = ['ID', 'source', 'feature', 'start', 'end', 'score', 'strand', 'frame', 'attribute'] # from gff format
repeat_gff['motif'] = repeat_gff['attribute'].str.extract('(?<=\:)(.*)(?=\")')
````

````python
repeat_gff
````

````python
# obtain a mapping table of motif to class
repeat_out = pd.read_table("../../output_t2t/sniffles/mosaic/multi_sample/repeatmasker/multi_sample.filtered.fa.out", header = None)
motif_family = repeat_out[0].str.strip().str.split("\s+", regex = True, expand = True)[[9, 10]].rename(columns = {9: 'motif', 10: 'family'}).drop_duplicates()
````

````python
# map class to output_gff motif
repeat_gff = repeat_gff.merge(motif_family, on = 'motif')
````

````python
# get counts of classes
family_count = pd.DataFrame(repeat_gff.groupby('family', as_index = False)['ID'].count().rename(columns = {'ID': 'count'}).sort_values('count', ascending = False))
# get counts of classes + motifs
motif_by_family_count = pd.DataFrame(repeat_gff.groupby(['family', 'motif'], as_index = False)['ID'].count().rename(columns = {'ID': 'count'}).sort_values('count', ascending = False))
````

````python
repeat_gff[repeat_gff['motif'] == 'Alu']
````

````python
fig, ax = plt.subplots(figsize = (6, 4))

sns.set_color_codes("pastel")
sns.barplot(x='family', y = 'count', data=family_count,
            label="Total", color="b")
sns.set_color_codes("muted")
sns.barplot(x='', y = 'count', data=family_count,
            label="Total", color="b")
# ax.bar(class_count.index, class_count.values)
ax.set_title("RepeatMasker annotation class counts")
plt.xticks(rotation=45, ha='right') 
plt.show()
````
