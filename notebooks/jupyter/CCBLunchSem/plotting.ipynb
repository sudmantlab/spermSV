```python
import os
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
from matplotlib import ticker
import seaborn as sns
import matplotlib_venn
```

```python
comp_table = pd.read_csv("cagan_bergeron_comparison_plotting.csv")
# enable arrangement of x axis by generation time
comp_table['species'] = pd.Categorical(comp_table['species'], ['Mouse', 'Dog', 'Tiger', 'Giraffe', 'Human'])
comp_table
```

```python
fig, ax = plt.subplots(figsize=(6,4))
sns.set_style()
sns.stripplot(ax = ax, data = comp_table, x = 'species', y = 'cagan_somatic_site_year', size = 8, label = 'Somatic (Cagan et al. 2022)')
sns.stripplot(ax = ax, data = comp_table, x = 'species', y = 'bergeron_germline_site_year', color = '#fab52a', size = 8, label = 'Germline (Bergeron et al. 2023)')
plt.ylabel("Mutation rate (site/year)")

# removes repetitive legend elements
h, l = ax.get_legend_handles_labels()
by_label = dict(zip(l,h))
plt.legend(by_label.values(), by_label.keys())

# add x to x 10-7 formatting
yfmt = ticker.ScalarFormatter(useMathText=True)
ax.yaxis.set_major_formatter(yfmt)

plt.xlabel("")
plt.title("Matched somatic & germline mutation rates", size = 16)
# sns.catplot(ax, data = comp_table, x = 'species', y = 'bergeron_germline_site_year')
```
